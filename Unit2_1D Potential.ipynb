```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
n=100
```

```python
A=np.zeros((n,n))
print(A)
```

```python
# buliding H matrix
h_bar= 1
m=1
C= -1 *((h_bar**2)/(2*m))
for i in range(n):
    for j in range(n):
        if i==j:
            A[i,j]=-2
        elif abs(i-j)== 1:
            A[i,j]=1       
print(A)
H=A*C
print(H)
```

**Particle in the Box**

```python
L = 1.0        # Box length
n = 100        # Grid points
dx = L / (n + 1)


# find egien value of matrix
energies, wavefunctions = np.linalg.eigh(H)

# egien vlaue
e_h=energies[0]
print(e_h)

#plot
x = np.linspace(0, L, n)
phi_0 = wavefunctions[:, 0]
plt.figure(figsize=(6, 3.5))
plt.plot(x, phi_0, label="Ground State Wave")
plt.title("Energy Level")
plt.xlabel("Box Position (x)")
plt.ylabel("Wave Height")
plt.legend()
plt.grid(True)
plt.show()
```

**Particle in finte potential wall**

```python
# we have H matrix
# define walls 
L=1.0 
a= 0.2
r= L/2 +a # right wall 
l= L/2 -a # left wall

# define potential
x = np.linspace(0, L, n)
v_0=.1
for i in range(n):
    if x[i]<(l) or x[i]>r:
        H[i,i]= C + v_0
    else:
        H[i,i]= C
# find egien value of matrix
energies, wavefunctions = np.linalg.eigh(H)

#plot
x = np.linspace(0, L, n)
phi_0 = wavefunctions[:, 0]
plt.plot(x, phi_0, label="Ground State Wave")
plt.title("Energy Level for finite well" )
plt.xlabel("Box Position (x)")
plt.ylabel("Wave Height")
plt.legend()
plt.grid(True)
plt.show()
```

**Harmonic potential**

```python
#define potential for harmonic
o= 1 # omega
x = np.linspace(-L, L, n)
for i in range(n):
    H[i,i]= H[i, i] + ((1/2)*x[i]**2)  # m =1 W =1
# find egien value of matrix
energies, wavefunctions = np.linalg.eigh(H)

#plot
x = np.linspace(-L, L, n)
phi_0 = wavefunctions[:, 0]
plt.plot(x, phi_0, label="Ground State Wave ")
plt.title("Energy Level for harmonic ")
plt.xlabel("Box Position (x)")
plt.ylabel("Wave Height")
plt.legend()
plt.grid(True)
plt.show()
```
