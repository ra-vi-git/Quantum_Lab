```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# H atom
n=5000
a_0 = 1 # atomic unit
# set r
#(R_min = 1e-15      # as v(0) is not defined)
R_max = 10 * a_0 
dr=R_max/n
# grid 
r = np.linspace(dr, R_max, n)
```

```python
# buliding H matrix

A=np.zeros((n,n))
h_bar= 1
m=1
C = -1 * ((h_bar**2) / (2 * m * (dr**2)))
for i in range(n):
    for j in range(n):
        if i==j:
            A[i,j]=-2
        elif abs(i-j)== 1:
            A[i,j]=1       
H=A*C
```

```python
# define potential
e = 1
# In atomic units, 4*pi*eps_0 is simply 1
V = np.diag((-e**2) / r)  # np.diag() make array to matix

# angular mometum term
l= 0 # angular momentum
D=np.diag(((h_bar**2)*l*(l+1))/(2*m*(r**2)))
```

```python
# final matix say F
F = H + V 
print(F)
```

**Ground State Radial Wavefunction**

```python
# find egien value of matrix
energies, wavefunctions = np.linalg.eigh(F)

#sloution
u_0 = -wavefunctions[:, 0]

#plot
plt.plot(r, u_0, label="Ground State Wave (1s)")
#plt.xlim(0,0.04)
plt.title("Hydrogen Atom Ground State Radial Wavefunction")
plt.xlabel("Radial Distance r (a.u)")
plt.ylabel("Wave Amplitude u(r)")
plt.legend()
plt.grid(True)
plt.show()
print(energies[0]*27.2114,"eV")  # 1 hartree = 27.2114 eV
```

**Frist excited State Radial Wavefunction**


```python
# find egien value of matrix
energies, wavefunctions = np.linalg.eigh(F)

#sloution
u_0 = -wavefunctions[:, 1]

#plot
plt.plot(r, u_0, label="Frist excited State Wave (2s)")
plt.title("Hydrogen Atom Ground State Radial Wavefunction")
plt.xlabel("Radial Distance r (a.u)")
plt.ylabel("Wave Amplitude u(r)")
plt.legend()
plt.grid(True)
plt.show()

print(energies[1]*27.2114,"eV")  # 1 hartree = 27.2114 eV
```

**Screened potential**

```python
# define screen potential
e = 1
a= 3*1.88973  # a = 3A, 1 A=1.88973 au
# In atomic units, 4*pi*eps_0 is simply 1
V_s = np.diag(((-e**2) / r)*((e**((-r)/a))))  # np.diag() make array to matix

# angular mometum term
l= 0 # angular momentum
D=np.diag(((h_bar**2)*l*(l+1))/(2*m*(r**2)))
```

```python
# final matix say F
F_s = H + V_s + D

# find egien value of matrix
energies_s, wavefunctions_s = np.linalg.eigh(F_s)

#sloution
u_0_s = -wavefunctions_s[:, 0]

#plot
plt.plot(r, u_0_s, label="Ground State Wave (1s)")
#plt.xlim(0,0.04)
plt.title("Hydrogen Atom Ground State Radial Wavefunction with screen potential")
plt.xlabel("Radial Distance r (a.u)")
plt.ylabel("Wave Amplitude u(r)")
plt.legend()
plt.grid(True)
plt.show()
print(energies_s[0]*27.2114,"eV")  # 1 hartree = 27.2114 eV

# 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
```
