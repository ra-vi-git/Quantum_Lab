```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# intial conditons 

n=5000
# set r
R_max = 10 #fm 
dr=R_max/(n+1)
# grid 
r = np.linspace(dr, R_max-dr, n)
```

```python
# define potential
m=940 #Mev/c^2
k=100 #MeV/fm^2
b=0
hbar_c=197.3 #MeV.fm
h_bar= 197.3
# In atomic units, 4*pi*eps_0 is simply 1
V = np.diag(((1/2)*k*(r**2)) + (1/3)*b*(r**2)) # np.diag() make array to matix

# angular mometum term
l= 0 # angular momentum
D=np.diag(((h_bar**2)*l*(l+1))/(2*m*(r**2)))
```

```python
# buliding H matrix

A=np.zeros((n,n))
h_bar= 197.3
C = -1 * ((h_bar**2) / (2 * m * (dr**2)))
for i in range(n):
    for j in range(n):
        if i==j:
            A[i,j]=-2
        elif abs(i-j)== 1:
            A[i,j]=1       
H=A*C
```

```python
# final matix say F
F = H + V + D 
print(F)
```

```python
# find egien value of matrix
energies, wavefunctions = np.linalg.eigh(F)

#sloution
u_0 = -wavefunctions[:, 0]

#plot
plt.plot(r, u_0, label="Ground State Wave (1s)")
#plt.xlim(0,0.04)
plt.title("Anharmonic oscillator ")
plt.xlabel("Radial Distance r (fm)")
plt.ylabel("Wave Amplitude u(r)")
plt.legend()
plt.grid(True)
plt.show()
print(energies[0],"MeV")
```

**Moderate Anharmonicity**

```python
b_m=10
V_m = np.diag(((1/2)*k*(r**2)) + (1/3)*b_m*(r**2))

# final matix say F
F_m = H + V_m + D 

energies, wavefunctions = np.linalg.eigh(F_m)

#sloution
u_0 = -wavefunctions[:, 0]

#plot
plt.plot(r, u_0, label="Ground State Wave (1s)")
#plt.xlim(0,0.04)
plt.title(" Moderate Anharmonic oscillator ")
plt.xlabel("Radial Distance r (fm)")
plt.ylabel("Wave Amplitude u(r)")
plt.legend()
plt.grid(True)
plt.show()
print(energies[0],"MeV")
```

**High Anharmonicity**

```python
b_h=30
V_h = np.diag(((1/2)*k*(r**2)) + (1/3)*b_h*(r**2))

# final matix say F
F_h = H + V_h + D 

energies, wavefunctions = np.linalg.eigh(F_h)

#sloution
u_0 = -wavefunctions[:, 1]

#plot
plt.plot(r, u_0, label="Ground State Wave (1s)")
#plt.xlim(0,0.04)
plt.title("High Anharmonic oscillator ")
plt.xlabel("Radial Distance r (fm)")
plt.ylabel("Wave Amplitude u(r)")
plt.legend()
plt.grid(True)
plt.show()
print(energies[1],"MeV")
```
